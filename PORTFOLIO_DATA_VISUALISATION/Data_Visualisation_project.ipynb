```python


from docx import Document
from docx.shared import Pt, Cm, RGBColor
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT, WD_ALIGN_VERTICAL
from docx.oxml.ns import qn
from docx.oxml import OxmlElement

def set_cell_border(cell):
    """Add border to all sides of a cell"""
    tc = cell._tc
    tcPr = tc.get_or_add_tcPr()
    for border_name in ['top', 'left', 'bottom', 'right']:
        border = OxmlElement(f'w:{border_name}')
        border.set(qn('w:val'), 'single')
        border.set(qn('w:sz'), '4')
        border.set(qn('w:space'), '0')
        border.set(qn('w:color'), '000000')
        tcPr.append(border)

def set_cell_background(cell, colour):
    """Set background colour of a cell"""
    tc = cell._tc
    tcPr = tc.get_or_add_tcPr()
    shd = OxmlElement('w:shd')
    shd.set(qn('w:val'), 'clear')
    shd.set(qn('w:color'), 'auto')
    shd.set(qn('w:fill'), colour)
    tcPr.append(shd)

# Create document
doc = Document()

# Set margins
for section in doc.sections:
    section.top_margin = Cm(2.54)
    section.bottom_margin = Cm(2.54)
    section.left_margin = Cm(2.54)
    section.right_margin = Cm(2.54)

# Add sentence above table
intro = doc.add_paragraph(
    'The dataset contains 7 variables, outlined below:')
intro.runs[0].font.size = Pt(12)
intro.runs[0].font.name = 'Times New Roman'

# Table data
table_data = [
    ['Variable', 'Description'],
    ['Date', 'The date on which the transaction occurred'],
    ['City', 'The city where the products are sold'],
    ['Confectionary', 'Types of product'],
    ['Units Sold', 'Quantity of items sold'],
    ['Revenue (£)', 'Total income generated'],
    ['Cost (£)', 'Expenses incurred'],
    ['Profit (£)', 'Net earnings (Revenue \u2013 Cost)'],
]

# Create table with 2 columns and 8 rows
table = doc.add_table(rows=8, cols=2)
table.alignment = WD_TABLE_ALIGNMENT.LEFT

# Set column widths
col_widths = [Cm(4), Cm(11)]
for i, width in enumerate(col_widths):
    for cell in table.columns[i].cells:
        cell.width = width

# Fill table data and apply formatting
for row_idx, row_data in enumerate(table_data):
    row = table.rows[row_idx]

    # Set row height
    row.height = Cm(0.85)

    for col_idx, text in enumerate(row_data):
        cell = row.cells[col_idx]

        # Clear existing content
        cell.text = ''

        # Add paragraph with text
        para = cell.paragraphs[0]
        para.alignment = WD_ALIGN_PARAGRAPH.CENTER
        run = para.add_run(text)
        run.font.name = 'Times New Roman'
        run.font.size = Pt(11)

        # Header row formatting
        if row_idx == 0:
            run.font.bold = True
            set_cell_background(cell, 'FFFFFF')
        else:
            run.font.bold = False
            set_cell_background(cell, 'FFFFFF')

        # Set vertical alignment to centre
        tc = cell._tc
        tcPr = tc.get_or_add_tcPr()
        vAlign = OxmlElement('w:vAlign')
        vAlign.set(qn('w:val'), 'center')
        tcPr.append(vAlign)

        # Add borders to cell
        set_cell_border(cell)

# Save the document
doc.save('Dataset_Variables_Table.docx')
print("SUCCESS: File saved as 'Dataset_Variables_Table.docx'")
print("Open the file, select the table, copy and paste")
print("it into your main report document.")
```
